```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

df = pd.read_csv("city_day.csv")
```

```python
print(df.head())

print(df.shape)

print(df.dtypes)
```

```python
print(df.describe())
```

```python
high_aqi = df[df["AQI"] > 300]

print(high_aqi.head())
```

```python
delhi = df[df["City"] == "Delhi"]

print(delhi.head())
```

```python
sorted_df = df.sort_values(
    by="AQI",
    ascending=False
)

print(sorted_df.head())
```

```python
print("Mean :", df["AQI"].mean())

print("Median :", df["AQI"].median())

print("Mode :")

print(df["AQI"].mode())

print("Standard Deviation :", df["AQI"].std())
```

```python
city_avg = df.groupby("City")["AQI"].mean()

print(city_avg.sort_values(ascending=False).head(10))
```

```python
city_avg.sort_values(ascending=False).head(10).plot(kind="bar")

plt.title("Top 10 Polluted Cities")

plt.ylabel("Average AQI")

plt.show()
```

```python
corr = df[["PM2.5","PM10","NO2","SO2","CO","AQI"]].corr()

print(corr)
```

```python
plt.figure(figsize=(8,6))

sns.heatmap(corr,
            annot=True,
            cmap="coolwarm")

plt.title("Correlation Heatmap")

plt.show()
```

```python
sns.scatterplot(data=df,
                x="PM2.5",
                y="AQI")

plt.title("PM2.5 vs AQI")

plt.show()
```
